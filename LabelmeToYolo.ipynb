#### Dependencies

#### Path and Objects

```python
images_path = os.path.join("C:/Users/Sina/Desktop/HamrahTel/Handwritten/yolo_dataset/images/")
labels_path = os.path.join("C:/Users/Sina/Desktop/HamrahTel/Handwritten/yolo_dataset/labels/")
images_list = os.listdir(images_path)
labels_list = os.listdir(labels_path)

# os.mkdir("C:/Users/Sina/Desktop/HamrahTel/Handwritten/yolo_dataset/text_labels/")
save_txt_path = os.path.join("C:/Users/Sina/Desktop/HamrahTel/Handwritten/yolo_dataset/text_labels/")
```

#### Loading json files (files which has been created using labelme)

```python
try:
    for i in range(len(labels_list)):
        with open(labels_path+labels_list[i], "r") as file:
            txt_data = []

            data = json.load(file)

            save_file_path = save_txt_path+labels_list[i].split(".")[0]+".txt"

            img_shape = cv.cvtColor(cv.imread(images_path+images_list[i]), cv.COLOR_BGR2GRAY).shape
            # shape[0] ---> height
            # shape[1] ---> width

            label_0 = data["shapes"][0]['label'] # --> 0:check
            label_1 = data["shapes"][1]['label'] # --> 1:area

            # --- check --- #
            x0_0 = int(data["shapes"][0]['points'][0][0])
            y0_0 = int(data["shapes"][0]['points'][0][1])
            x1_0 = int(data["shapes"][0]['points'][1][0])
            y1_0 = int(data["shapes"][0]['points'][1][1])
            x_center_0 = ((x0_0 + x1_0)/2) / img_shape[1] # Normalization has been applied directly
            y_center_0 = ((y0_0 + y1_0)/2) / img_shape[0] # Normalization has been applied directly

            width_0 = abs(x1_0 - x0_0) / img_shape[1] # Width and Height Normalization
            height_0 = abs(y1_0 - y0_0) / img_shape[0] # Width and Height Normalization

            # --- area --- #
            x0_1 = int(data["shapes"][1]['points'][0][0])
            y0_1 = int(data["shapes"][1]['points'][0][1])
            x1_1 = int(data["shapes"][1]['points'][1][0])
            y1_1 = int(data["shapes"][1]['points'][1][1])
            x_center_1 = ((x0_1 + x1_1)/2) / img_shape[1] # Normalization has been applied directly
            y_center_1 = ((y0_1 + y1_1)/2) / img_shape[0] # Normalization has been applied directly

            width_1 = abs(x1_1 - x0_1) / img_shape[1] # Width and Height Normalization
            height_1 = abs(y1_1 - y0_1) / img_shape[0] # Width and Height Normalization

            txt_data_0 = ["0", x_center_0, y_center_0, width_0, height_0]
            txt_data_1 = ["1", x_center_1, y_center_1, width_1, height_1]

            with open(save_file_path, 'w') as f:
                f.write(" ".join(map(str, txt_data_0)) + "\n")
                f.write(" ".join(map(str, txt_data_1)) + "\n")

            print(f"data number '{i}' has been saved")
            
except Exception as e:
    print(f"corrupted data '{i}' ---> \n{e}")
```

#### Resizing and Converting images to Gray and 640*640

```python
old_images = "old_images"
images_path = os.path.join(f"C:/Users/Sina/Desktop/HamrahTel/Handwritten/yolo_dataset/{old_images}/")
images_list = os.listdir(images_path)

for i in range(len(images_list)):
    save_path = "C:/Users/Sina/Desktop/HamrahTel/Handwritten/yolo_dataset/images/"+images_list[i]
    
    img = cv.cvtColor(cv.imread(images_path+images_list[i]), cv.COLOR_BGR2GRAY)
    new_img = cv.resize(img, (640,640))
    cv.imwrite(save_path, new_img)
```
